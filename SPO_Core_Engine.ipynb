```python
# Smart Process Optimizer (SPO)
# KPI Input and Analysis Engine

# Step 1 - Define the KPIs
kpi_data = {
    "efficiency_rate": 65,
    "cycle_time": 45,
    "waste_percentage": 12,
    "roi": 8,
    "manpower_utilization": 70,
    "rejection_rate": 8,
    "lead_time": 10
}

print("KPI Data Loaded Successfully!")
print(kpi_data)
```

```python
# Step 2 - Industry Benchmarks
# These are the standard values a healthy manufacturing plant should have

benchmarks = {
    "efficiency_rate": 85,      # 85% and above is good
    "cycle_time": 30,           # 30 seconds/minutes is the target
    "waste_percentage": 5,      # 5% or below is good
    "roi": 15,                  # 15% ROI is the target
    "manpower_utilization": 85, # 85% and above is good
    "rejection_rate": 3,        # 3% or below is good
    "lead_time": 7              # 7 days or below is good
}

print("Industry Benchmarks Loaded Successfully!")
print(benchmarks)
```

```python
# Step 3 - Analysis Engine
# Compare plant KPIs against industry benchmarks

def analyze_kpis(kpi_data, benchmarks):
    results = {}
    
    for kpi, value in kpi_data.items():
        benchmark = benchmarks[kpi]
        
        # KPIs where higher is better
        if kpi in ["efficiency_rate", "roi", "manpower_utilization"]:
            gap = value - benchmark
            if gap >= 0:
                status = "Good"
            elif gap >= -10:
                status = "Needs Improvement"
            else:
                status = "Critical"
                
        # KPIs where lower is better
        else:
            gap = benchmark - value
            if gap >= 0:
                status = "Good"
            elif gap >= -10:
                status = "Needs Improvement"
            else:
                status = "Critical"
        
        results[kpi] = {
            "value": value,
            "benchmark": benchmark,
            "gap": abs(gap),
            "status": status
        }
    
    return results

analysis = analyze_kpis(kpi_data, benchmarks)

for kpi, result in analysis.items():
    print(f"{kpi}: {result['value']} | Benchmark: {result['benchmark']} | Status: {result['status']}")
```

```python
# Step 4 - Root Cause Suggester
# Based on underperforming KPIs, suggest probable root causes

def suggest_root_causes(analysis):
    root_causes = []

    if analysis["efficiency_rate"]["status"] in ["Needs Improvement", "Critical"]:
        root_causes.append("Low efficiency detected - Possible causes: machine downtime, poor line balancing, or excessive idle time")

    if analysis["cycle_time"]["status"] in ["Needs Improvement", "Critical"]:
        root_causes.append("High cycle time detected - Possible causes: bottleneck at a specific station, unskilled labor, or poor workstation layout")

    if analysis["waste_percentage"]["status"] in ["Needs Improvement", "Critical"]:
        root_causes.append("High waste detected - Possible causes: overproduction, defective materials, or poor quality control")

    if analysis["roi"]["status"] in ["Needs Improvement", "Critical"]:
        root_causes.append("Low ROI detected - Possible causes: high operational costs, low output, or high rejection rates")

    if analysis["manpower_utilization"]["status"] in ["Needs Improvement", "Critical"]:
        root_causes.append("Low manpower utilization detected - Possible causes: poor task distribution, idle workers, or line imbalance")

    if analysis["rejection_rate"]["status"] in ["Needs Improvement", "Critical"]:
        root_causes.append("High rejection rate detected - Possible causes: quality control issues, defective raw materials, or operator errors")

    if analysis["lead_time"]["status"] in ["Needs Improvement", "Critical"]:
        root_causes.append("High lead time detected - Possible causes: supply chain delays, poor scheduling, or production bottlenecks")

    return root_causes

root_causes = suggest_root_causes(analysis)

print("ROOT CAUSE ANALYSIS")
print("=" * 50)
for i, cause in enumerate(root_causes, 1):
    print(f"{i}. {cause}")
```

```python
# Step 5 - Recommendations Engine
# Based on root causes, suggest specific actionable recommendations

def generate_recommendations(analysis):
    recommendations = []

    if analysis["efficiency_rate"]["status"] in ["Needs Improvement", "Critical"]:
        recommendations.append("Conduct a full line balancing study to redistribute workload evenly across all stations")

    if analysis["cycle_time"]["status"] in ["Needs Improvement", "Critical"]:
        recommendations.append("Perform time studies at each station to identify the bottleneck and redistribute tasks accordingly")

    if analysis["waste_percentage"]["status"] in ["Needs Improvement", "Critical"]:
        recommendations.append("Implement a waste tracking system and investigate the top 3 sources of waste on the production floor")

    if analysis["roi"]["status"] in ["Needs Improvement", "Critical"]:
        recommendations.append("Focus on reducing operational costs by eliminating non value adding activities across the production process")

    if analysis["manpower_utilization"]["status"] in ["Needs Improvement", "Critical"]:
        recommendations.append("Reassign idle workers as floaters to support overloaded stations or assign them to value adding support tasks")

    if analysis["rejection_rate"]["status"] in ["Needs Improvement", "Critical"]:
        recommendations.append("Conduct root cause analysis on top defect types and implement quality checkpoints at critical production stages")

    if analysis["lead_time"]["status"] in ["Needs Improvement", "Critical"]:
        recommendations.append("Map the entire value stream to identify delays and implement pull based scheduling to reduce lead time")

    return recommendations

recommendations = generate_recommendations(analysis)

print("RECOMMENDATIONS")
print("=" * 50)
for i, rec in enumerate(recommendations, 1):
    print(f"{i}. {rec}")
```

```python
# Step 6 - What-If Simulator (Fixed)

def what_if_simulator(kpi_data, improvements):
    print("WHAT-IF SIMULATION RESULTS")
    print("=" * 50)
    
    for kpi, improvement in improvements.items():
        current = kpi_data[kpi]
        projected = current + improvement
        
        # Show + or - correctly
        if improvement > 0:
            change = f"+{improvement}"
        else:
            change = f"{improvement}"
            
        print(f"{kpi}: Current: {current} | Projected: {projected} | Change: {change}")

# Test scenario
improvements = {
    "efficiency_rate": 12,
    "manpower_utilization": 15,
    "rejection_rate": -5,
    "waste_percentage": -4,
}

what_if_simulator(kpi_data, improvements)
```
